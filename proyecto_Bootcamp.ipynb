```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.neural_network import MLPClassifier
import numpy as np
```

```python
def recomendacion_rapida():
    """Función interactiva para obtener recomendaciones"""

    # Cargar datos
    print("📂 Cargando base de datos de asociaciones...")
    df = pd.read_csv('/content/ASOCIACIONES_DE_PRODUCTORES_AGROPECUARIOS_DEPARTAMENTO_DE_BOYACÀ_20251003.csv')
    df.columns = df.columns.str.strip()

    # Mostrar opciones disponibles
    municipios_disponibles = df['Municipio'].unique()
    productos_disponibles = df['Producto'].unique()

    print(f"\n📍 Municipios disponibles: {len(municipios_disponibles)}")
    print(f"🌱 Productos disponibles: {len(productos_disponibles)}")

    # Solicitar inputs al usuario
    print("\n" + "="*50)
    print("🎯 BUSCADOR DE ASOCIACIONES PARA ALIANZAS")
    print("="*50)

    municipio = input("\n🏘️  Ingresa el MUNICIPIO: ").strip().title()
    producto = input("🌱 Ingresa el PRODUCTO: ").strip().title()

    print(f"\n🔍 Buscando asociaciones en {municipio} para {producto}...")

    # Filtrar por municipio y producto
    resultados = df[
        (df['Municipio'].str.contains(municipio, case=False, na=False)) &
        (df['Producto'].str.contains(producto, case=False, na=False)) &
        (df['Estado de la Asociacion'].str.contains('Activa', na=False))
    ]

    if not resultados.empty:
        print(f"\n✅ Se encontraron {len(resultados)} asociaciones:")
        print("-" * 70)

        for idx, row in resultados.head(10).iterrows():
            print(f"\n🏆 {row['Nombre de la Asociación']}")
            print(f"   📍 Municipio: {row['Municipio']}")
            print(f"   🌱 Producto: {row['Producto']}")
            print(f"   👥 Asociados activos: {row['Número de asociados  Activos']}")
            print(f"   🏢 Estado: {row['Estado de la Asociacion']}")
            print(f"   🆔 NIT: {row['Nit']}")

            # Información adicional útil para alianzas
            if pd.notna(row['La Asociación cuenta con una sede']) and row['La Asociación cuenta con una sede'] == 'Si':
                print(f"   🏛️  Cuenta con sede propia")
            if pd.notna(row['La asociación cuenta con centro de acopio']) and row['La asociación cuenta con centro de acopio'] == 'Si':
                print(f"   📦 Cuenta con centro de acopio")
            if pd.notna(row['Tienen una cuenta bancaria, ahorro, corriente ?.']) and row['Tienen una cuenta bancaria, ahorro, corriente ?.'] == 'Si':
                print(f"   💰 Cuenta bancaria activa")

    else:
        print("\n❌ No se encontraron asociaciones exactas.")

        # Búsqueda flexible - solo por producto
        resultados_flexibles = df[
            (df['Producto'].str.contains(producto, case=False, na=False)) &
            (df['Estado de la Asociacion'].str.contains('Activa', na=False))
        ]

        if not resultados_flexibles.empty:
            print(f"\n💡 Te sugerimos estas asociaciones de {producto} en otros municipios:")
            for idx, row in resultados_flexibles.head(5).iterrows():
                print(f"   • {row['Nombre de la Asociación']} - {row['Municipio']}")

        # Búsqueda flexible - solo por municipio
        resultados_municipio = df[
            (df['Municipio'].str.contains(municipio, case=False, na=False)) &
            (df['Estado de la Asociacion'].str.contains('Activa', na=False))
        ]

        if not resultados_municipio.empty:
            print(f"\n💡 Asociaciones activas en {municipio}:")
            productos_municipio = resultados_municipio['Producto'].unique()[:5]
            for prod in productos_municipio:
                print(f"   • {prod}")

# Función para mostrar todas las opciones disponibles
def explorar_opciones():
    """Muestra todas las opciones disponibles al usuario"""

    df = pd.read_csv('/content/ASOCIACIONES_DE_PRODUCTORES_AGROPECUARIOS_DEPARTAMENTO_DE_BOYACÀ_20251003.csv')
    df.columns = df.columns.str.strip()

    print("\n📊 EXPLORAR OPCIONES DISPONIBLES")
    print("="*40)

    municipios = df['Municipio'].astype(str).unique()
    productos = df['Producto'].astype(str).unique()

    print(f"\n📍 MUNICIPIOS ({len(municipios)}):")
    for i, mun in enumerate(sorted(municipios)[:20]):  # Mostrar primeros 20
        print(f"   {i+1}. {mun}")
    if len(municipios) > 20:
        print(f"   ... y {len(municipios)-20} más")

    print(f"\n🌱 PRODUCTOS ({len(productos)}):")
    for i, prod in enumerate(sorted(productos)[:20]):  # Mostrar primeros 20
        print(f"   {i+1}. {prod}")
    if len(productos) > 20:
        print(f"   ... y {len(productos)-20} más")

# Menú principal interactivo
def menu_principal():
    while True:
        print("\n" + "="*60)
        print("🤝 SISTEMA DE ALIANZAS CON ASOCIACIONES AGROPECUARIAS - BOYACÁ")
        print("="*60)
        print("1. 🔍 Buscar asociaciones por municipio y producto")
        print("2. 📊 Explorar opciones disponibles")
        print("3. ❌ Salir")

        opcion = input("\nSelecciona una opción (1-3): ").strip()

        if opcion == '1':
            recomendacion_rapida()
        elif opcion == '2':
            explorar_opciones()
        elif opcion == '3':
            print("¡Hasta pronto! 👋")
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2 o 3.")

# Ejecutar el programa
if __name__ == "__main__":
    menu_principal()
```
